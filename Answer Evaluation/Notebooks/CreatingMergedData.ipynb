```python
scores_long = scores.melt(id_vars=["Question"],
                  value_vars=["R1", "R2", "R3", "R4", "R5", "R6", "R7", "R8", "R9", "R10"],
                  var_name="Response_Number",
                  value_name="Score")
```

```python
responses_df = responses_long.drop_duplicates(subset=["Question", "Response_Number"])
scores_df = scores_long.drop_duplicates(subset=["Question", "Response_Number"])
```

```python
merged_df = pd.merge(responses_df, scores_df, on=["Question", "Response_Number"])
```

```python
print("Duplicate rows in responses:", responses_df.duplicated(subset=["Question", "Response_Number"]).sum())
print("Duplicate rows in scores:", scores_df.duplicated(subset=["Question", "Response_Number"]).sum())
print("Duplicate rows in Merged DF:", merged_df.duplicated(subset=["Question", "Response_Number"]).sum())
```

```python
# Merge correct answers into the long-form dataframe
final_df = pd.merge(merged_df, df, on="Question", how="left")


```
