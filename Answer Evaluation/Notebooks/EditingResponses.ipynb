```python
import pandas as pd
```

```python
# Load your existing datasets
df_responses = pd.read_csv("responses (1).csv")
df_scores = pd.read_csv("scores (1).csv")
```

```python
def update_responses_and_scores(responses, scores, row_index, save=False):
    """
    Update responses and scores for a specific question row.

    Parameters:
    - responses: list of 10 strings
    - scores: list of 10 floats (between 0 and 1)
    - row_index: int, row number in dataset
    - save: bool, if True, saves updated CSVs
    """

    # Safety checks
    assert len(responses) == 10, "Must provide 10 responses."
    assert len(scores) == 10, "Must provide 10 scores."

    for i in range(10):
        df_responses.loc[row_index, f"R{i+1}"] = responses[i]
        df_scores.loc[row_index, f"R{i+1}"] = scores[i]

    print(f"Updated row {row_index} with responses and scores.")
```
